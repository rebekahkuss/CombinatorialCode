1.1) Suppose you have six different marbles, labeled 1 through 6, in a bag. You pick two different marbles out of the bag and put them in your pocket. Write out all of the possible ways you could have two different marbles in your pocket. (You don’t want to count picking marbles 1 and 2 and picking marbles 2 and 1 as different outcomes).

1.2) Suppose you have six different marbles, labeled 1 through 6, in a bag. You pick three different marbles out of the bag and put them in your pocket. Write out all of the possible ways you could have three different marbles in your pocket. (You don’t want to count picking marbles 1, 2, and 3, and picking marbles 3, 2, and 1 as different outcomes).
