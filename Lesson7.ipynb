$\Large \textbf{Lesson 7 - Task 1}$

1.1) How many arrangements are there of the letters in the word MAMMA? First, list all of the outcomes by hand. Then write some code to list all of these outcomes.

Are there other representations (tree diagrams, etc.) that help you reason about how to solve this problem?


1.2) How many arrangements are there of the letters in the word BANANA? Again, write some code to list all possible outcomes. 

1.3)	How many 6-digit binary sequences (sequences of just 0s and 1s) contain exactly 4 1s? Exactly 3 1s? Write some code to list all of the outcomes.

1.4)	How many 6-digit ternary sequences (sequences of just 0s, 1s, and 2s) contain exactly 4 1s? Exactly 3 1s? Write some code to list all of the outcomes.
