## Understanding the Math

### Permutations Formula
$$P(n,r) = \frac{n!}{(n-r)!}$$

Use this when **order matters**. For example, the arrangement ABC is different from BAC.

### Combinations Formula  
$$C(n,r) = \frac{n!}{r!(n-r)!}$$

Use this when **order doesn't matter**. For example, the selection {A,B,C} is the same as {C,B,A}.

### Counting Principle
If there are $m$ ways to do one thing and $n$ ways to do another thing, then there are $m \times n$ ways to do both.

This extends to any number of independent choices: $n_1 \times n_2 \times n_3 \times ... \times n_k$

```python
# Generate actual permutations and combinations
from itertools import permutations, combinations

items = ['A', 'B', 'C', 'D']

print("Permutations of 2 items from ['A', 'B', 'C', 'D']:")
for p in permutations(items, 2):
    print(p)

print("\nCombinations of 2 items from ['A', 'B', 'C', 'D']:")
for c in combinations(items, 2):
    print(c)
```
