$\Large \textbf{Task 2}$

2.1) A standard deck of cards has 52 cards, 13 of each suit: Hearts, Spades, Clubs, and Diamonds. Consider the Spade-Heart Problem: How many possible ways are there to pick a pair of cards if the first card is a spade and the second card is a heart? 

Can you write code that will solve this counting problem and list all of the possible combinations?

2.2) Now consider the Face-Heart Problem How many possible ways are there to pick a pair of cards (without replacement) if the first card is a face and the second card is a heart? 

Can you write code that will solve this counting problem and list all of the possible combinations?
